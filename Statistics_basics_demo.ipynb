```python
import pandas as pd
sample_data = {'name': ['John', 'Alia', 'Ananya','Steve','vish'],
               'gender': ['M','F','F','M','M'],
               'communication_skill_score': [40,45,23,39,39],
               'quantative_skill_score':[38,41,42,48,32]}
data = pd.DataFrame(sample_data)
data
```

```python
print('Mean of Communication skill score: ',data['communication_skill_score'].mean())
print('Mean of Quantative skill score:',data['quantative_skill_score'].mean())
```

```python
print('Mode of Communication skill score: ', data['communication_skill_score'].mode())
print('Mode of Quantitative skill score: ', data['quantative_skill_score'].mode())
```

```python
print('Median of Communication skill score: ', data['communication_skill_score'].median())
print('Median of Quantitative skill score: ', data['quantative_skill_score'].median())
```

```python
q1 = data['communication_skill_score'].quantile(.25)
q3 = data['communication_skill_score'].quantile(.75)
iqr = q3 - q1
print('Inter Quartile Range of Communication skill score: ', iqr)
```

```python
communication_skill_range = data['communication_skill_score'].max() - data['communication_skill_score'].min()
quantative_skill_range = data['quantative_skill_score'].max() - data['quantative_skill_score'].min()
print('Range of Communication skill score: ', communication_skill_range)
print('Range of Quantative skill score: ', quantative_skill_range)
```

```python
communication_skill_variance = data['communication_skill_score'].var()
quantative_skill_variance = data['quantative_skill_score'].var()
print('Variance of Communication skill score: ', communication_skill_variance)
print('Variance of Quantative skill score: ', quantative_skill_variance)
```

```python
communication_skill_std = data['communication_skill_score'].std()
quantative_skill_std = data['quantative_skill_score'].std()
print('Standard Deviation of Communication skill score: ', communication_skill_std)
print('Standard Deviation of Quantative skill score: ', quantative_skill_std)
```

```python
data.describe()
```

```python
communication_skill_kurt = data['communication_skill_score'].kurtosis()
quantative_skill_kurt = data['quantative_skill_score'].kurtosis()
print('Kurtosis of Communication skill score: ', communication_skill_kurt)
print('Kurtosis of Quantative skill score: ', quantative_skill_kurt)
```

```python
# Covariance of data
print('Covariance of Communication skill score and Quantative skill score: ', data['communication_skill_score'].cov(data['quantative_skill_score']))
print('Covariance of Quantative skill score and Communication skill score: ', data['quantative_skill_score'].cov(data['communication_skill_score']))
```

```python
# Corelation of communication skill score
print('Corelation of Communication skill score and Quantative skill score: ', data['communication_skill_score'].corr(data['quantative_skill_score']))
```
